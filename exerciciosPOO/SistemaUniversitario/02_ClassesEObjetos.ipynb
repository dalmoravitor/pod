### **Exercício: Sistema de Gestão Universitária**  

Você deverá criar um sistema de gestão universitária com três classes `Professor`, `Departamento` e `Curso`.
Seu objetivo será implementar as classes e testar os seus funcionamentos.

---

### **Requisitos das Classes**  

#### **1. Classe `Professor`**  
**Atributos:**  
- `nome` (string)  
- `especialidade` (string)  

**Métodos:**  
- `get_info()` → Retorna string formatada: `"Nome: [nome], Especialidade: [especialidade]"`.  
- `atualizar_especialidade(nova_especialidade)` → Altera a especialidade.  

**Métodos Especiais/Sobrecarga:**  
- `__eq__(self, other)` → Compara dois professores pelo **nome** (retorna `True` se forem iguais).  
- `__str__(self)` → Retorna `"Professor: [nome]"`.  

---

```python
class Professor:

    def __init__(self, nome, especialidade):
        self.nome = nome
        self.especialidade = especialidade


    def get_info(self):
        return f"Nome: {self.nome}, especialidade: {self.especialidade}"

    def  __str__(self):
        return f"Professor: {self.nome}"
```

#### **2. Classe `Departamento`**  
**Atributos:**  
- `nome` (string)  
- `chefe` (instância de `Professor`, inicialmente `None`)  

**Métodos:**  
- `alterar_chefe(novo_chefe)` →  
  - Valida se `novo_chefe` é instância de `Professor`.  
  - Retorna mensagens de sucesso/erro.  
- `get_info()` → Retorna string com nome do departamento e info do chefe.  

**Métodos Especiais/Sobrecarga:**  
- `__add__(self, other)` →  
  - **Sobrecarga do operador `+`**: Cria um novo departamento fictício com nome combinado (ex: `depto1 + depto2` → nome `"Ciência da Computação & Matemática"`).  
- `__str__(self)` → Retorna `"Departamento: [nome]"`.  

```python
class Departamento:
    def __init__(self, nome, chefe = None):
        self.nome = nome
        self.chefe = chefe

    def alterar_chefe(self, novo_chefe):
        if novo_chefe.isinstance(Professor):
            self.chefe = novo_chefe
            return f"Sucesso!"
        else:
            return f"Erro: O novo chefe não é instância da classe professor"

    def get_info(self):
        return f"Nome do departamento: {self.nome}, chefe: {self.chefe}"


    def __str__(self):
        return f"Departamento: {self.nome}"
        
```

#### **3. Classe `Curso`**  
**Atributos:**  
- `nome` (string)  
- `departamento` (instância de `Departamento`)  
- `carga_horaria` (int, entre 30 e 200)  

**Métodos:**  
- `atualizar_carga_horaria(nova_carga)` → Valida e atualiza a carga horária.  
- `get_info()` → Retorna string com nome do curso, carga horária e departamento.  

**Métodos Especiais/Sobrecarga:**  
- `__lt__(self, other)` → Compara cursos pela carga horária (retorna `True` se `self.carga_horaria < other.carga_horaria`).
- `__le__(self, other)` → Compara cursos pela carga horária (retorna `True` se `self.carga_horaria <= other.carga_horaria`).
- `__gt__(self, other)` → Compara cursos pela carga horária (retorna `True` se `self.carga_horaria > other.carga_horaria`).
- `__ge__(self, other)` → Compara cursos pela carga horária (retorna `True` se `self.carga_horaria >= other.carga_horaria`).
- `__ne__(self, other)` → Compara cursos pelo **nome** (retorna `True` se forem diferentes).
- `__eq__(self, other)` → Compara cursos pelo **nome** (retorna `True` se forem iguais).
- `__str__(self)` → Retorna `"Curso: [nome]"`.  

```python
class Curso:

    def __init__(self, nome, departamento, carga_horaria):
        self.nome = nome
        self.departamento = departamento
        self.carga_horaria = carga_horaria

    def atualizar_carga_horaria(self, nova_carga):
        if nova_carga >= 30 and nova_carga <= 200:
            self.nova_carga = nova_carga
            return f"Sucesso: carga horaria atualizada para {nova_carga}."
        else:
            return f"Erro: a nova carga precisa ter entre 30 a 200 horas."

    def get_info(self):
        return f"Nome do curso: {self.nome}, departamento: {self.departamento}, carga horária: {self.carga_horaria}"


    def __str__(self):
        return f"Curso: {self.nome}"

    
```
