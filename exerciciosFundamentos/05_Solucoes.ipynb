```python
# 1
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print(f"Conjunto união: {A.union(B)}")
print(f"Conjunto interceção: {A.intersection(B)}")
print(f"Conjunto diferença: {A.difference(B)}")
print(f"A é subconjunto de B?: {A.issubset(B)}")

```

```python
dicionario1 = {"a": 1, "b": 2}
dicionario2 = {"b": 3, "c": 4}

mesclado = {}

for chave in dicionario1:
    if chave in dicionario2:
        mesclado[chave] = dicionario1[chave] + dicionario2[chave]
    else:
        mesclado[chave] = dicionario1[chave]
for chave in dicionario2:
    if chave not in mesclado:
        mesclado[chave] = dicionario2[chave]
print(mesclado)
```
