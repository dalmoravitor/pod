## 1. Gere um método que receba de entrada do usuário dois números e retorne a soma.
- **Descrição**: Crie uma função que receba dois números como parâmetros e retorne a soma desses números.
- **Exemplo de Entrada**: `3, 5`
- **Exemplo de Saída**: `8`

## 2. Gere um método que receba de entrada do usuário um nome e uma mensagem de saudação, e retorne a mensagem personalizada.
- **Descrição**: Crie uma função que receba um nome e uma mensagem de saudação (com um valor default) e retorne a mensagem personalizada.
- **Exemplo de Entrada**: `"João", "Bom dia"`
- **Exemplo de Saída**: `"Bom dia, João!"`

## 3. Gere um método que receba de entrada do usuário uma quantidade variável de números e retorne a soma deles.
- **Descrição**: Crie uma função que receba uma quantidade variável de números usando `*args` e retorne a soma desses números.
- **Exemplo de Entrada**: `3, 5, 7, 2`
- **Exemplo de Saída**: `17`

## 4. Gere um método que receba de entrada do usuário informações sobre um produto (nome, preço, quantidade) e retorne um resumo formatado.
- **Descrição**: Crie uma função que receba informações sobre um produto usando `**kwargs` e retorne um resumo formatado.
- **Exemplo de Entrada**: `nome="Caneta", preço=1.5, quantidade=10`
- **Exemplo de Saída**: `"Produto: Caneta, Preço: 1.5, Quantidade: 10"`

## 5. Gere um método que receba de entrada do usuário um número e retorne o quadrado e o cubo desse número.
- **Descrição**: Crie uma função que receba um número e retorne o quadrado e o cubo desse número.
- **Exemplo de Entrada**: `3`
- **Exemplo de Saída**: `(9, 27)`

## 6. Gere um método que recebe uma quantidade qualquer de números e retorne a média deles.
- **Descrição**: Crie uma função que receba uma quantidade variável de números usando `*args` e retorne a média desses números.
- **Exemplo de Entrada**: `3, 5, 7, 9`
- **Exemplo de Saída**: `6.0`

## 7. Gere um método que recebe uma string e retorna a mesma string invertida.
- **Descrição**: Crie uma função que receba uma string e retorne a mesma string invertida.
- **Exemplo de Entrada**: `"Python"`
- **Exemplo de Saída**: `"nohtyP"`

## 8. Gere um método que recebe uma quantidade qualquer de números e retorna a soma dos números pares.
- **Descrição**: Crie uma função que receba uma quantidade variável de números usando `*args` e retorne a soma dos números pares.
- **Exemplo de Entrada**: `1, 2, 3, 4, 5`
- **Exemplo de Saída**: `6` (2 + 4)

## 9. Gere um método que recebe um inteiro e retorna o fatorial desse número.
- **Descrição**: Crie uma função que receba um número inteiro e retorne o fatorial desse número.
- **Exemplo de Entrada**: `5`
- **Exemplo de Saída**: `120` (5! = 5 * 4 * 3 * 2 * 1)

## 10. Gere um método que recebe uma string e retorna a contagem de vogais e consoantes.
- **Descrição**: Crie uma função que receba uma string e retorne a contagem de vogais e consoantes em um único retorno.
- **Exemplo de Entrada**: `"Python"`
- **Exemplo de Saída**: `(1, 5)` (1 vogal e 5 consoantes)

## 11. Crie um método que recebe um numero e retorna se ele é primo ou não.
- **Descrição**: Crie uma função que receba um número e retorne se ele é primo ou não.
- **Exemplo de Entrada**: `7`
- **Exemplo de Saída**: `True` (7 é primo)

## 12. Crie um método que recebe uma string e um número e retorna a string repetida o número de vezes informado.
- **Descrição**: Crie uma função que receba uma string e um número e retorne a string repetida o número de vezes informado.
- **Exemplo de Entrada**: `"Python", 3`
- **Exemplo de Saída**: `"PythonPythonPython"`
